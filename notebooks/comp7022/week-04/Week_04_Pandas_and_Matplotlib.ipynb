## Lecture 10 | Pandas Basics

### Exercise 01 · Create a DataFrame

**Task:** Create a table with Name, Age, and Score for five students.

**Solution origin:** reviewed/reconstructed reference.


```python
import pandas as pd
# Each dictionary key becomes a DataFrame column.
students = {
    "Name": ["Alan", "Sarah", "John", "Emma", "David"],
    "Age": [23, 21, 25, 22, 24],
    "Score": [75, 90, 85, 95, 70],
}
df = pd.DataFrame(students)
print("Original student DataFrame:")
print(df)
```

**Explanation:** The dictionary values become aligned columns; `pd.DataFrame` builds a labelled table.

### Exercise 02 · Select one or multiple columns

**Task:** Select Score, then select Name and Age together.

**Solution origin:** reviewed/reconstructed reference.


```python
# One bracket returns a Series; a list of names returns a DataFrame.
print("Scores:")
print(df["Score"])
print("Names and ages:")
print(df[["Name", "Age"]])
```

**Explanation:** `df["Score"]` is a Series; `df[["Name", "Age"]]` is a DataFrame.

### Exercise 03 · Select a row with iloc

**Task:** Use a numeric position to select the fourth student.

**Solution origin:** reviewed/reconstructed reference.


```python
# Position 3 refers to the fourth row because positions start at zero.
print("Fourth student:")
print(df.iloc[3])
```

**Explanation:** `iloc` uses positions, which start at zero.

### Exercise 04 · Label-based indexing with loc

**Task:** Make Name the index; find Sarah and John's score.

**Solution origin:** reviewed/reconstructed reference.


```python
named_df = df.set_index("Name")
print("Sarah:")
print(named_df.loc["Sarah"])
print("John's score:")
print(named_df.loc["John", "Score"])
```

**Explanation:** `loc` selects label values; `set_index` returns a new table unless assigned back.

### Exercise 05 · Filter scores above 85

**Task:** Keep only rows with scores greater than 85.

**Solution origin:** reviewed/reconstructed reference.


```python
# The comparison creates a Boolean mask applied to the rows.
print("Scores above 85:")
print(df[df["Score"] > 85])
```

**Explanation:** `>` excludes the value 85 itself.

### Exercise 06 · Combine conditions with AND

**Task:** Keep students younger than 25 with scores above 80.

**Solution origin:** reviewed/reconstructed reference.


```python
filtered_and = df[(df["Age"] < 25) & (df["Score"] > 80)]
print("Age under 25 and score above 80:")
print(filtered_and)
```

**Explanation:** Use `&` and parentheses around each comparison.

### Exercise 07 · Combine conditions with OR

**Task:** Keep students younger than 22 OR with scores at least 85.

**Solution origin:** reviewed/reconstructed reference.


```python
filtered_or = df[(df["Age"] < 22) | (df["Score"] >= 85)]
print("Age under 22 or score at least 85:")
print(filtered_or)
```

**Explanation:** `|` accepts either condition. `>=` also includes equality.

### Exercise 08 · Combine two inclusive conditions

**Task:** Find students aged 22 or older AND scoring at least 85.

**Solution origin:** reviewed/reconstructed reference.


```python
filtered_both = df[(df["Age"] >= 22) & (df["Score"] >= 85)]
print("Age at least 22 and score at least 85:")
print(filtered_both)
```

**Explanation:** Two inclusive conditions must both hold because the operator is `&`.

### Exercise 09 · Inspect DataFrame structure

**Task:** Inspect columns, data types and metadata.

**Solution origin:** reviewed/reconstructed reference.


```python
# columns and dtypes are attributes; info() is a method.
print("Columns:", df.columns.tolist())
print("Dtypes:")
print(df.dtypes)
print("DataFrame information:")
df.info()
```

**Explanation:** Attributes have no `()`; methods such as `info()` are called with parentheses.

### Exercise 10 · Describe and calculate the mean

**Task:** Print numeric summaries and calculate the mean score.

**Solution origin:** reviewed/reconstructed reference.


```python
print("Numeric summary:")
print(df.describe())
print("Average score:", df["Score"].mean())
```

**Explanation:** `describe()` summarizes numeric columns and `mean()` computes the arithmetic average.

### Exercise 11 · Write and reload CSV

**Task:** Save the student data and reload it without retaining the DataFrame index.

**Solution origin:** reviewed/reconstructed reference.


```python
from tempfile import TemporaryDirectory
from pathlib import Path
# Use a temporary folder to avoid adding generated files to the portfolio.
with TemporaryDirectory() as temp_dir:
    csv_path = Path(temp_dir) / "students.csv"
    df.to_csv(csv_path, index=False)
    loaded_df = pd.read_csv(csv_path)
    print("Loaded CSV data:")
    print(loaded_df.head())
```

**Explanation:** `index=False` avoids saving a redundant numeric index; `read_csv` restores columns.

## Lecture 11 | DataFrame Manipulation

### Exercise 14 · Create a calculated column

**Task:** Create DoubleScore by multiplying Score by two.

**Solution origin:** reviewed/reconstructed reference.


```python
df["DoubleScore"] = df["Score"] * 2
print("Scores and doubled scores:")
print(df[["Name", "Score", "DoubleScore"]])
```

**Explanation:** Pandas performs vectorized multiplication on the entire column.

### Exercise 15 · Mean across numeric columns

**Task:** Calculate means for all numeric columns, skipping text columns.

**Solution origin:** reviewed/reconstructed reference.


```python
print("Numeric column averages:")
print(df.mean(numeric_only=True))
```

**Explanation:** `numeric_only=True` excludes columns such as Name.

### Exercise 16 · Apply a function row-wise

**Task:** Use apply to add Score and DoubleScore within each row.

**Solution origin:** reviewed/reconstructed reference.


```python
numeric_scores = df[["Score", "DoubleScore"]]
row_totals = numeric_scores.apply(sum, axis=1)
print("Score totals by student:")
print(row_totals)
```

**Explanation:** `axis=1` operates across columns within each row.

### Exercise 17 · Apply a custom function column-wise

**Task:** Create a function that doubles the sum of each numeric column; use axis=0.

**Solution origin:** reviewed/reconstructed reference.


```python
def calculate_double_total(values):
    # Return the double of the sum of the supplied values.
    total = sum(values)
    return total * 2

numeric_df = df[["Age", "Score", "DoubleScore"]]
result = numeric_df.apply(calculate_double_total, axis=0)
print("Double totals for numeric columns:")
print(result)
```

**Explanation:** `axis=0` feeds each column to the function; `return` sends back its result.

### Exercise 18 · Map Boolean values with lambda

**Task:** Create HighScore and map it to Excellent or Other.

**Solution origin:** reviewed/reconstructed reference.


```python
df["HighScore"] = df["Score"] >= 90
# map applies the lambda to each Boolean value in the Series.
df["Result"] = df["HighScore"].map(lambda x: "Excellent" if x else "Other")
print("Score labels:")
print(df[["Name", "Score", "Result"]])
```

**Explanation:** `lambda` selects a text label for each Boolean value.

### Exercise 19 · Center the Age column

**Task:** Subtract the mean Age to create Age_Centered; verify its mean.

**Solution origin:** reviewed study-session solution.


```python
# Original solution from our study practice.
df["Age_Centered"] = df["Age"] - df["Age"].mean()
print(df[["Name", "Age", "Age_Centered"]])
print("Centered age mean:", df["Age_Centered"].mean())
```

**Explanation:** The original age values differ from the mean by -2, -1, 0, 1, or 2; centering is not full standardization.

### Exercise 20 · Standardize the Score column

**Task:** Create Score_Standardized using mean and standard deviation; inspect both properties.

**Solution origin:** reviewed study-session solution.


```python
# A z-score is measured in units of standard deviations.
df["Score_Standardized"] = (df["Score"] - df["Score"].mean()) / df["Score"].std()
print(df[["Name", "Score", "Score_Standardized"]])
print("Standardized score mean:", df["Score_Standardized"].mean())
print("Standardized score std:", df["Score_Standardized"].std())
```

**Explanation:** Centering makes the mean zero; division by `std()` also makes the sample standard deviation one.

### Exercise 21 · Calculate percentage change

**Task:** Create a Change column for Close versus Open; calculate average percentage change.

**Solution origin:** reviewed study-session solution.


```python
prices = pd.DataFrame({
    "Day": ["Monday", "Tuesday", "Wednesday", "Thursday"],
    "Open": [100, 120, 90, 80],
    "Close": [110, 114, 99, 76],
})
prices["Change"] = 100 * (prices["Close"] - prices["Open"]) / prices["Open"]
print(prices[["Day", "Open", "Close", "Change"]])
print("Average percentage change:", prices["Change"].mean())
```

**Explanation:** A negative change means Close is below Open; the mean here is +2.5 percentage points across the four changes.

### Exercise 22 · Boolean analysis of price growth

**Task:** Mark whether Close is higher than Open and count days of growth.

**Solution origin:** reviewed study-session solution.


```python
# True if the closing price exceeds the opening price.
prices["Grew"] = prices["Close"] > prices["Open"]
print(prices[["Day", "Open", "Close", "Change", "Grew"]])
print("Days with price increases:", prices["Grew"].sum())
```

**Explanation:** `True` counts as one in a Boolean sum; unchanged prices produce `False`.

### Exercise 23 · Sort records by percent change

**Task:** Sort sales by Change_Percent from largest to smallest.

**Solution origin:** reviewed study-session solution.


```python
sales = pd.DataFrame({
    "Product": ["Laptop", "Monitor", "Keyboard", "Mouse"],
    "Sales_2025": [100, 80, 200, 150],
    "Sales_2026": [120, 60, 250, 150],
})
sales["Change_Percent"] = 100 * (sales["Sales_2026"] - sales["Sales_2025"]) / sales["Sales_2025"]
sorted_sales = sales.sort_values("Change_Percent", ascending=False)
print(sorted_sales[["Product", "Change_Percent"]])
```

**Explanation:** `sort_values` reorders rows but preserves their original index labels.

### Exercise 24 · Reset the sorted index

**Task:** Renumber sorted rows from zero without keeping the old index column.

**Solution origin:** reviewed study-session solution.


```python
sorted_sales = sorted_sales.reset_index(drop=True)
print(sorted_sales[["Product", "Change_Percent"]])
```

**Explanation:** `drop=True` discards old index labels; reassignment saves the new DataFrame.

### Exercise 28 · Concatenate two DataFrames

**Task:** Combine annual order tables, using new unique sequential indexes.

**Solution origin:** reviewed study-session solution.


```python
orders_2025 = pd.DataFrame({"Product": ["Laptop", "Monitor"], "Units": [10, 5]})
orders_2026 = pd.DataFrame({"Product": ["Keyboard", "Mouse"], "Units": [15, 20]})
all_orders = pd.concat([orders_2025, orders_2026], ignore_index=True)
print(all_orders)
# Without ignore_index=True, duplicate labels 0 and 1 remain.
with_duplicate_index = pd.concat([orders_2025, orders_2026])
print("Rows labelled 1 without index reset:")
print(with_duplicate_index.loc[1])
```

### Exercise 35 · Combine Pandas groupby with Matplotlib

**Task:** Sum Units by Product and pass the Series index and values to plt.bar().

**Solution origin:** reviewed study-session solution.


```python
sales_data = pd.DataFrame({
    "Product": ["Laptop", "Monitor", "Keyboard", "Laptop", "Monitor", "Keyboard", "Laptop", "Monitor"],
    "Units": [10, 5, 20, 15, 8, 25, 12, 7],
})
total_sales = sales_data.groupby("Product")["Units"].sum()
print("Total units by product:")
print(total_sales)
# Series.index holds product names; Series.values holds totals.
plt.bar(total_sales.index, total_sales.values, color="green")
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Units Sold")
plt.show()
```

**Explanation:** The aggregation avoids drawing several overlapping bars for repeated product labels.

### Final task 1 · Boolean filtering

**Task:** Create a Boolean column Increased for growth from 2025 to 2026, filter True rows, and count them.

```python
sales_df = pd.DataFrame({
    "Product": ["Laptop", "Monitor", "Keyboard", "Mouse", "Laptop", "Monitor"],
    "Category": ["Computers", "Displays", "Accessories", "Accessories", "Computers", "Displays"],
    "Units_2025": [10, 8, 15, 20, 12, 5],
    "Units_2026": [14, 7, 18, 25, 15, 9],
})
sales_df["Increased"] = sales_df["Units_2025"] < sales_df["Units_2026"]
increased_sales = sales_df[sales_df["Increased"] == True]
print(increased_sales[["Product", "Units_2025", "Units_2026", "Increased"]])
print("Records with increased sales:", increased_sales["Increased"].sum())
```

**Review:** The Boolean column marks all rows; filtering uses the mask to select a subset. There are five growing records.

### Final task 2 · Percentage change and filtering

**Task:** Calculate Change_Percent, average it, and retain records growing by more than 25%.

```python
sales_df["Change_Percent"] = 100 * (sales_df["Units_2026"] - sales_df["Units_2025"]) / sales_df["Units_2025"]
print(sales_df[["Product", "Units_2025", "Units_2026", "Change_Percent"]])
print("Average percentage change:", sales_df["Change_Percent"].mean())
high_growth = sales_df[sales_df["Change_Percent"] > 25]
print(high_growth[["Product", "Change_Percent"]])
```

**Review:** The >25 condition excludes exactly 25%. A negative percentage is a decline from the 2025 baseline.

### Final task 3 · Group, aggregate and sort

**Task:** Summarize 2026 Units by Category with sum and mean, sorted by sum descending.

```python
category_summary = sales_df.groupby("Category")["Units_2026"].agg(["sum", "mean"])
category_summary = category_summary.sort_values(["sum"], ascending=False)
print(category_summary)
```

**Review:** Each category appears once, with separate sum and mean columns. Accessories 43/21.5; Computers 29/14.5; Displays 16/8.0.

### Final task 4 · Center and standardize

**Task:** Compute centered and standardized 2026 units; verify mean and std.

```python
sales_df["Units_Centered"] = sales_df["Units_2026"] - sales_df["Units_2026"].mean()
print("Original units mean:", sales_df["Units_2026"].mean())
print(sales_df[["Product", "Units_2026", "Units_Centered"]])
sales_df["Units_Standardized"] = sales_df["Units_Centered"] / sales_df["Units_2026"].std()
print(sales_df[["Product", "Units_2026", "Units_Standardized"]])
print("Standardized mean:", sales_df["Units_Standardized"].mean())
print("Standardized std:", sales_df["Units_Standardized"].std())
```

**Review:** Centered -5 means five units below the mean, whereas standardized -1.2 means 1.2 standard deviations below it.

### Final task 5 · Concatenation and index reset

**Task:** Concatenate two order tables without ignore_index, sort units descending, then reset the index.

```python
orders_a = pd.DataFrame({"Product": ["Laptop", "Monitor", "Keyboard"], "Units": [15, 8, 25]})
orders_b = pd.DataFrame({"Product": ["Mouse", "Printer", "Tablet"], "Units": [30, 12, 20]})
all_orders = pd.concat([orders_a, orders_b])
all_orders = all_orders.sort_values(["Units"], ascending=False)
all_orders = all_orders.reset_index(drop=True)
print(all_orders)
```

**Review:** The original concatenation retained duplicate labels 0, 1, 2; the reset creates fresh unique labels 0–5.
