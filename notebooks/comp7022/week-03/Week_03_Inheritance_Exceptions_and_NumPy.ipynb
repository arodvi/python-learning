## Lecture 7 – Inheritance & OOP Extensions

### Exercise 01 · Basic inheritance

**Task.** Create a subclass that inherits a parent method.

**Why it works.** A child class inherits suitable methods from its parent.

*Solution status: reviewed/reconstructed reference example.*

```python
class Vehicle:
    def move(self):
        return "Vehicle is moving"

class Car(Vehicle):
    pass

car = Car()
print("Move:", car.move())
```

### Exercise 02 · Method overriding and super

**Task.** Extend inherited behavior while invoking the parent method.

**Why it works.** super() gives access to the parent implementation.

*Solution status: reviewed/reconstructed reference example.*

```python
class Animal:
    def speak(self):
        return "Animal sound"

class Dog(Animal):
    def speak(self):
        return super().speak() + " -> Woof!"

print("Dog sound:", Dog().speak())
```

### Exercise 03 · Parent constructor

**Task.** Call super().__init__() from a child class.

**Why it works.** The parent initializer sets inherited instance attributes.

*Solution status: reviewed/reconstructed reference example.*

```python
class Person:
    def __init__(self, name):
        self.name = name

class UniversityStudent(Person):
    def __init__(self, name, student_id):
        super().__init__(name)
        self.student_id = student_id

student = UniversityStudent("Alan", "K100")
print("Student:", student.name, student.student_id)
```

### Exercise 04 · Properties and setters

**Task.** Validate assignments using @property and a setter.

**Why it works.** Property setters can reject invalid values without exposing internal state.

*Solution status: reviewed/reconstructed reference example.*

```python
class GradeBook:
    def __init__(self):
        self._grade = 60

    @property
    def grade(self):
        return self._grade

    @grade.setter
    def grade(self, value):
        if 0 <= value <= 100:
            self._grade = value

g = GradeBook()
g.grade = 150
print("Grade after invalid assignment:", g.grade)
```

### Exercise 05 · Multiple inheritance

**Task.** Combine methods from more than one base class.

**Why it works.** Multiple inheritance requires care around method resolution order.

*Solution status: reviewed/reconstructed reference example.*

```python
class Teacher:
    def teach(self):
        return "Teaching"

class Researcher:
    def research(self):
        return "Researching"

class Professor(Teacher, Researcher):
    pass

professor = Professor()
print("Teaching:", professor.teach())
print("Research:", professor.research())
```

### Exercise 06 · Abstract base classes

**Task.** Implement an abstract method in a concrete subclass.

**Why it works.** Instantiating a class with unimplemented abstract methods raises TypeError.

*Solution status: reviewed/reconstructed reference example.*

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class RectangleShape(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    def area(self):
        return self.width * self.height

shape = RectangleShape(3, 5)
print("Rectangle area:", shape.area())
```

### Exercise 07 · isinstance and issubclass

**Task.** Check objects and types against inheritance relationships.

**Why it works.** These two functions answer different kinds of type questions.

*Solution status: reviewed/reconstructed reference example.*

```python
print("Is shape a Shape:", isinstance(shape, Shape))
print("Is RectangleShape a Shape subclass:", issubclass(RectangleShape, Shape))
print("Is Shape a RectangleShape subclass:", issubclass(Shape, RectangleShape))
```

*Solution status: reviewed/reconstructed reference example.*

```python
print("Exact type:", type(shape).__name__)
print("Has area method:", "area" in dir(shape))
```

## Lecture 8 – Exceptions & NumPy Basics

### Exercise 09 · ValueError versus TypeError

**Task.** Identify the difference between invalid values and incompatible types.

**Why it works.** Exceptions communicate different failure categories.

*Solution status: reviewed/reconstructed reference example.*

```python
try:
    int("hello")
except ValueError as error:
    print("ValueError caught:", type(error).__name__)

try:
    3 + "4"
except TypeError as error:
    print("TypeError caught:", type(error).__name__)
```

### Exercise 10 · try and except

**Task.** Catch an expected exception without stopping the entire program.

**Why it works.** Catch the most specific suitable exception.

*Solution status: reviewed/reconstructed reference example.*

```python
try:
    number = int("not a number")
    print("Converted number:", number)
except ValueError:
    print("Invalid integer input")
```

### Exercise 11 · else and finally

**Task.** Observe the control flow of complete exception handling.

**Why it works.** else runs on success; finally runs regardless.

*Solution status: reviewed/reconstructed reference example.*

```python
try:
    value = int("42")
except ValueError:
    print("Conversion failed")
else:
    print("Conversion succeeded:", value)
finally:
    print("Conversion attempt finished")
```

### Exercise 12 · Raising an exception

**Task.** Reject invalid function input with raise.

**Why it works.** raise interrupts normal execution and transfers control to a handler.

*Solution status: reviewed/reconstructed reference example.*

```python
def double_non_negative(x):
    if x < 0:
        raise ValueError("Negative value not allowed")
    return x * 2

try:
    print("Result:", double_non_negative(-3))
except ValueError:
    print("Invalid value")
```

### Exercise 25 · NumPy binary I/O

**Task.** Save and reopen an ndarray in NumPy binary format.

**Why it works.** The .npy format preserves dtype and shape.

*Solution status: reviewed/reconstructed reference example.*

```python
from pathlib import Path
from tempfile import TemporaryDirectory
import numpy as np

with TemporaryDirectory() as temporary_dir:
    path = Path(temporary_dir) / "student_grades.npy"
    grades = np.array([55, 65, 75, 85, 95])
    np.save(path, grades)
    loaded_grades = np.load(path)
    print("Loaded NPY grades:", loaded_grades)
    print("Loaded dtype:", loaded_grades.dtype)
```

### Exercise 26 · NumPy CSV I/O

**Task.** Save and reopen numeric values in CSV format.

**Why it works.** Text loaders commonly infer float unless given dtype explicitly.

*Solution status: reviewed/reconstructed reference example.*

```python
from pathlib import Path
from tempfile import TemporaryDirectory
import numpy as np

with TemporaryDirectory() as temporary_dir:
    path = Path(temporary_dir) / "student_grades.csv"
    grades = np.array([55, 65, 75, 85, 95])
    np.savetxt(path, grades, delimiter=",")
    loaded_grades = np.loadtxt(path, delimiter=",")
    print("Loaded CSV grades:", loaded_grades)
    print("Mean grade:", loaded_grades.mean())
```

## Knowledge check · NumPy and OOP

**1. What does super().show() do in an overridden method?**

*Answer:* Invokes the parent implementation before returning to the child method.

**2. What happens if a subclass does not implement an abstract method?**

*Answer:* Instantiating it raises TypeError.

**3. What happens when a property setter silently rejects an invalid value?**

*Answer:* The original attribute value remains unchanged.

**4. Which blocks run after a ValueError caught in except?**

*Answer:* The matching except and finally blocks; else is skipped.

**5. Does raise allow the following return statement to run?**

*Answer:* No. It transfers control by raising an exception.

**6. What does np.arange(1,13).reshape(3,4)[1,2] evaluate to?**

*Answer:* 7.

**7. What does np.array([[2,4,6],[1,3,5]]).sum(axis=0) return?**

*Answer:* [3, 7, 11].

**8. Can you combine masks with & and persist arrays with np.save?**

*Answer:* Yes; & is elementwise AND and np.save uses a binary NumPy format.

**9. What does np.array([10,20,30,40])[[3,1,1]] select?**

*Answer:* [40, 20, 20].

**10. What is np.dot([[1,2],[3,4]], [[2,0],[1,2]])?**

*Answer:* [[4, 4], [10, 8]].
