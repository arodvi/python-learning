### Exercise 03 · for loops and range

**Task.** Enumerate a collection and generate even numbers.

**Why it works.** range excludes its stop value; enumerate supplies indexes.

*Solution status: reviewed/reconstructed reference example.*

```python
subjects = ["Python", "AI", "Data Science"]
for index, subject in enumerate(subjects, start=1):
    print("Subject", index, ":", subject)

print("Even numbers:", list(range(2, 11, 2)))
```

### Exercise 04 · while loops and break

**Task.** Accumulate five numbers using a loop that terminates explicitly.

**Why it works.** A break statement exits the loop immediately.

*Solution status: reviewed/reconstructed reference example.*

```python
count = 0
total = 0
while True:
    count += 1
    if count > 5:
        break
    total += count
print("Sum from 1 to 5:", total)
```

### Exercise 05 · Basic descriptive statistics

**Task.** Calculate an average and count passing scores.

**Why it works.** sum and len can produce descriptive statistics without external packages.

*Solution status: reviewed/reconstructed reference example.*

```python
python_scores = [65, 78, 90, 45, 82, 55, 95]
average = round(sum(python_scores) / len(python_scores), 2)
passed_count = sum(1 for score in python_scores if score >= 60)
print("Average:", average)
print("Passed students:", passed_count)
print("Highest score:", max(python_scores))
```

**Task.** Construct a value mapping and a set of unique remainders.

**Why it works.** Comprehensions can create several kinds of collection.

*Solution status: reviewed/reconstructed reference example.*

### Exercise 16 · Nested comprehension

**Why it works.** Two for clauses traverse outer rows then elements.

*Solution status: reviewed/reconstructed reference example.*

### Exercise 17 · Sequence slicing

**Task.** Slice a list by start, stop and step.

*Solution status: reviewed/reconstructed reference example.*

```python
topics = ["Intro", "Python", "AI", "Data Science", "ML", "Cloud"]
print("Middle topics:", topics[1:4])
print("Every second topic:", topics[::2])
print("Last two topics:", topics[-2:])
print("Reversed:", topics[::-1])
```

### Exercise 18 · Average calculator

**Task.** Implement a reusable average function.

**Why it works.** Function parameters should be used instead of relying on external variables.

*Solution status: reviewed/reconstructed reference example.*

```python
def calculate_average(numbers):
    if not numbers:
        return None
    return sum(numbers) / len(numbers)

python_marks = [7, 8, 6, 9]
print("Average mark:", calculate_average(python_marks))
```

### Practical 05 · Class 2 · Task 5: tic-tac-toe

**Task.** Detect a winning row, column or diagonal in a 3×3 game.

**Source context:** adapted from course practice. This is a reference implementation, not claimed as an original submission.

```python
def winning_player(board):
    lines = list(board)
    lines += [[board[row][col] for row in range(3)] for col in range(3)]
    lines += [[board[i][i] for i in range(3)]]
    lines += [[board[i][2 - i] for i in range(3)]]
    for line in lines:
        if line[0] in {"x", "o"} and line[0] == line[1] == line[2]:
            return line[0]
    return "none"

board = [["x", "o", "o"], ["o", "x", "x"], ["x", "x", "o"]]
print("Winner:", winning_player(board))
assert winning_player([["x"] * 3, ["o", "x", "o"], ["o", "o", "x"]]) == "x"
assert winning_player(board) == "none"
```

### Practical 08 · Class 3 · Task 3: hash table

**Task.** Build a minimal list-of-buckets key-value store, including collision handling and overwrites.

**Source context:** adapted from course practice. This is a reference implementation, not claimed as an original submission.

```python
class BucketMap:
    def __init__(self, capacity=10):
        self.buckets = [[] for _ in range(capacity)]

    def _index(self, key):
        return hash(key) % len(self.buckets)

    def put(self, key, value):
        bucket = self.buckets[self._index(key)]
        for position, (stored_key, _) in enumerate(bucket):
            if stored_key == key:
                bucket[position] = (key, value)
                return
        bucket.append((key, value))

    def get(self, key, default=None):
        for stored_key, value in self.buckets[self._index(key)]:
            if stored_key == key:
                return value
        return default

store = BucketMap(capacity=2)
store.put("alan", "7943")
store.put("sarah", "2711")
store.put("alan", "8000")
print("Alan value:", store.get("alan"))
print("Missing key:", store.get("missing"))
assert store.get("alan") == "8000"
```
