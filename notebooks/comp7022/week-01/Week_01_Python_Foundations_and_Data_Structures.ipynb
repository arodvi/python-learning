### Exercise 04 · while loops and break

**Task.** Accumulate five numbers using a loop that terminates explicitly.

**Why it works.** A break statement exits the loop immediately.

*Solution status: reviewed/reconstructed reference example.*

````python
# Initialize a total and counter before the loop.
count = 0
# while True starts an otherwise unbounded loop.
total = 0
# The break condition stops the loop before adding a sixth value.
while True:
    count += 1
    if count > 5:
        break
    total += count
print("Sum from 1 to 5:", total)
````


**Step-by-step explanation**

1. Initialize a total and counter before the loop.

**Common mistake to avoid:** Without the break condition this loop would never finish.

### Practical 05 · Class 2 · Task 5: tic-tac-toe

**Task.** Detect a winning row, column or diagonal in a 3×3 game.

**Source context:** adapted from course practice. This is a reference implementation, not claimed as an original submission.

````python
# Each row, column and diagonal forms a potential winning line.
def winning_player(board):
    lines = list(board)
    lines += [[board[row][col] for row in range(3)] for col in range(3)]
    lines += [[board[i][i] for i in range(3)]]
    lines += [[board[i][2 - i] for i in range(3)]]
    for line in lines:
        if line[0] in {"x", "o"} and line[0] == line[1] == line[2]:
            return line[0]
    return "none"

# The function collects all eight possible lines of a 3×3 board.
board = [["x", "o", "o"], ["o", "x", "x"], ["x", "x", "o"]]
print("Winner:", winning_player(board))
# It returns the marker only when all three cells match a valid player.
assert winning_player([["x"] * 3, ["o", "x", "o"], ["o", "o", "x"]]) == "x"
assert winning_player(board) == "none"
````

**Observed output (example run)**

```text
Winner: none
```

**Step-by-step explanation**

1. Each row, column and diagonal forms a potential winning line.
2. The function collects all eight possible lines of a 3×3 board.
3. It returns the marker only when all three cells match a valid player.

**Common mistake to avoid:** Without the valid-player check, three empty cells could be mistaken for a win.

### Practical 08 · Class 3 · Task 3: hash table

**Task.** Build a minimal list-of-buckets key-value store, including collision handling and overwrites.

**Source context:** adapted from course practice. This is a reference implementation, not claimed as an original submission.

````python
# hash(key) maps a key to a number; modulo maps that number to one of the bucket indexes.
class BucketMap:
    def __init__(self, capacity=10):
        self.buckets = [[] for _ in range(capacity)]

    def _index(self, key):
        return hash(key) % len(self.buckets)

    def put(self, key, value):
        bucket = self.buckets[self._index(key)]
        for position, (stored_key, _) in enumerate(bucket):
            if stored_key == key:
                bucket[position] = (key, value)
                return
        bucket.append((key, value))

    def get(self, key, default=None):
        for stored_key, value in self.buckets[self._index(key)]:
            if stored_key == key:
                return value
        return default

store = BucketMap(capacity=2)
# Each bucket is a list, so colliding keys can coexist in that list.
store.put("alan", "7943")
store.put("sarah", "2711")
store.put("alan", "8000")
# put updates an existing key instead of adding a duplicate; get checks equality and returns the matching value.
print("Alan value:", store.get("alan"))
print("Missing key:", store.get("missing"))
assert store.get("alan") == "8000"
````

**Observed output (example run)**

```text
Alan value: 8000
Missing key: None
```

**Step-by-step explanation**

1. hash(key) maps a key to a number; modulo maps that number to one of the bucket indexes.
2. Each bucket is a list, so colliding keys can coexist in that list.
3. put updates an existing key instead of adding a duplicate; get checks equality and returns the matching value.

**Common mistake to avoid:** A hash collision is expected behavior; never identify records solely by equal hash values.
