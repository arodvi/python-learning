## Lecture 4 – Functions

### Exercise 01 · Functions and return values

**Task.** Create a function that returns the average of supplied numbers.

**Why it works.** return passes a value to the caller.

*Solution status: reviewed/reconstructed reference example.*

```python
def calculate_average(numbers):
    return sum(numbers) / len(numbers)

marks = [7, 8, 6, 9]
average = calculate_average(marks)
print("Average:", average)
```

### Exercise 02 · print versus return

**Task.** Use a function result in another calculation.

**Why it works.** Displaying a value and returning it are different operations.

*Solution status: reviewed/reconstructed reference example.*

```python
def add_numbers(a, b):
    return a + b

result = add_numbers(5, 3)
print("Result:", result)
print("Doubled result:", result * 2)
```

### Exercise 03 · Positional and keyword arguments

**Task.** Use default parameters and supply optional arguments.

**Why it works.** Keyword arguments name parameters; defaults fill omitted values.

*Solution status: reviewed/reconstructed reference example.*

### Exercise 06 · Local and global scope

**Task.** Read a global value inside a function without changing it.

**Why it works.** Local names remain inside their function scope.

*Solution status: reviewed/reconstructed reference example.*

```python
multiplier = 3

def multiply_value(value):
    local_result = value * multiplier
    return local_result

print("Product:", multiply_value(7))
```

### Exercise 07 · Modifying a global variable

**Task.** Explain and demonstrate the global keyword.

**Why it works.** global is needed when reassigning a module-level name.

*Solution status: reviewed/reconstructed reference example.*

```python
counter = 0

def increase_counter():
    global counter
    counter += 1

increase_counter()
increase_counter()
print("Global counter:", counter)
```

### Exercise 08 · Closures and nonlocal

**Task.** Update a variable captured from an enclosing function.

**Why it works.** nonlocal allows reassignment in an enclosing non-global scope.

*Solution status: reviewed/reconstructed reference example.*

```python
def create_counter():
    count = 0
    def next_count():
        nonlocal count
        count += 1
        return count
    return next_count

next_number = create_counter()
print("Call 1:", next_number())
print("Call 2:", next_number())
```

### Exercise 09 · Multiple return values

**Why it works.** Comma-separated return values form a tuple.

*Solution status: reviewed/reconstructed reference example.*

```python
def limits(values):
    return min(values), max(values)

minimum, maximum = limits([4, 8, 2, 10])
print("Minimum:", minimum)
print("Maximum:", maximum)
```

### Exercise 10 · Recursion

**Task.** Write a recursive function with a base case.

**Why it works.** A base case is essential to terminate recursion.

*Solution status: reviewed/reconstructed reference example.*

```python
def factorial(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n == 0:
        return 1
    return n * factorial(n - 1)

print("Factorial 5:", factorial(5))
```

## Lecture 6 – Object-Oriented Python

### Exercise 17 · Defining a class

**Task.** Create a class and instantiate an object.

**Why it works.** Each instance has its own identity.

*Solution status: reviewed/reconstructed reference example.*

```python
class Student:
    def __init__(self, name, mark):
        self.name = name
        self.mark = mark

alan = Student("Alan", 85)
sarah = Student("Sarah", 92)
print("First student:", alan.name, alan.mark)
print("Second student:", sarah.name, sarah.mark)
```

### Exercise 18 · Methods and instance state

**Task.** Update instance attributes through a method.

**Why it works.** self refers to the current instance.

*Solution status: reviewed/reconstructed reference example.*

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount

    def get_balance(self):
        return self.balance

account = BankAccount("Sally", 100)
account.deposit(50)
print("Owner:", account.owner)
print("Balance:", account.get_balance())
```

### Exercise 19 · Class versus instance attributes

**Task.** Compare attributes shared by a class to attributes stored on an instance.

**Why it works.** A class attribute may be shared; assignments on instances can shadow it.

*Solution status: reviewed/reconstructed reference example.*

```python
class CourseStudent:
    course = "COMP7022"
    def __init__(self, name):
        self.name = name

a = CourseStudent("Alan")
b = CourseStudent("Emma")
print("Course:", a.course, b.course)
print("Names:", a.name, b.name)
```

### Exercise 20 · Method composition

**Task.** Call one instance method from another.

**Why it works.** Instance methods can reuse behavior through self.method().

*Solution status: reviewed/reconstructed reference example.*

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def report(self):
        return f"Area: {self.area()}"

rectangle = Rectangle(4, 6)
print(rectangle.report())
```

### Practical 01 · Class 5 · Task 1: recursive Fibonacci

**Task.** Use three parameters (two with defaults) to calculate the eighth Fibonacci value.

**Source context:** adapted from course practice. This is a reference implementation, not claimed as an original submission.

```python
def fibonacci(position, previous=0, current=1):
    if position < 1:
        raise ValueError("Position must be positive")
    if position == 1:
        return previous
    return fibonacci(position - 1, current, previous + current)

print("Fibonacci position 8:", fibonacci(8))
assert fibonacci(8) == 13
```

### Practical 05 · Class 5 · Task 5: factorial with reduce

**Task.** Compute factorial using reduce, and verify it against the standard library.

**Source context:** adapted from course practice. This is a reference implementation, not claimed as an original submission.

```python
from functools import reduce
from math import factorial
from operator import mul

def factorial_reduce(n):
    if n < 0:
        raise ValueError("Factorial is undefined for negative integers")
    return reduce(mul, range(1, n + 1), 1)

for n in (10, 17, 22, 31):
    computed = factorial_reduce(n)
    assert computed == factorial(n)
    print("Factorial validated for:", n)
```

### Practical 06 · Class 5 · Task 6: higher-order function

**Task.** Write a helper that accepts an operation and its data, then returns a list.

**Source context:** adapted from course practice. This is a reference implementation, not claimed as an original submission.

```python
def convolve(operation, function, data):
    return list(operation(function, data))

groups = [[1, 5, 3], [3, 6], [9, 11, 45, 11], [5, 2, 6, 3, 7]]
result = convolve(map, lambda group: sum(group), groups)
print("Convolved sums:", result)
assert result == [9, 9, 76, 23]
```
