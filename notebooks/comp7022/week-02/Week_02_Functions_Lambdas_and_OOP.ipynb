### Exercise 01 · Functions and return values

**Task.** Create a function that returns the average of supplied numbers.

**Why it works.** return passes a value to the caller.

*Solution status: reviewed/reconstructed reference example.*

````python
# The function takes a list via its numbers argument.
def calculate_average(numbers):
    return sum(numbers) / len(numbers)

# sum(numbers) calculates the numerator; len(numbers) supplies the denominator.
marks = [7, 8, 6, 9]
# return sends the average to the assignment, after which print displays it.
average = calculate_average(marks)
print("Average:", average)
````

**Observed output (example run)**

```text
Average: 7.5
```

**Step-by-step explanation**

1. The function takes a list via its numbers argument.
2. sum(numbers) calculates the numerator; len(numbers) supplies the denominator.
3. return sends the average to the assignment, after which print displays it.

**Common mistake to avoid:** An empty list would cause ZeroDivisionError unless validated.

### Exercise 06 · Local and global scope

*Solution status: reviewed/reconstructed reference example.*

````python
# multiplier is created in the module scope.
multiplier = 3

# multiply_value reads the module-level multiplier and creates a local result.
def multiply_value(value):
    local_result = value * multiplier
    return local_result

# return passes the local value outside; the local name itself is not available there.
print("Product:", multiply_value(7))
````

### Exercise 07 · Modifying a global variable

**Task.** Explain and demonstrate the global keyword.

**Why it works.** global is needed when reassigning a module-level name.

*Solution status: reviewed/reconstructed reference example.*

````python
# counter starts at module scope.
counter = 0

# The global declaration makes assignment target that outer name.
def increase_counter():
    global counter
    counter += 1

increase_counter()
# Each call adds one to the same counter, so calling twice prints two.
increase_counter()
print("Global counter:", counter)
````

**Observed output (example run)**

```text
Global counter: 2
```

**Step-by-step explanation**

1. counter starts at module scope.
2. The global declaration makes assignment target that outer name.

**Common mistake to avoid:** Global state can make functions harder to test; use it only deliberately.

### Exercise 08 · Closures and nonlocal

**Task.** Update a variable captured from an enclosing function.

**Why it works.** nonlocal allows reassignment in an enclosing non-global scope.

*Solution status: reviewed/reconstructed reference example.*

````python
# create_counter creates its own count and a nested function.
def create_counter():
    count = 0
    def next_count():
        nonlocal count
        count += 1
        return count
    return next_count

# nonlocal tells next_count to update count from its enclosing function.
next_number = create_counter()
# Returning next_count preserves access to that state for later calls.
print("Call 1:", next_number())
print("Call 2:", next_number())
````

**Observed output (example run)**

```text
Call 1: 1
Call 2: 2
```

**Step-by-step explanation**

1. create_counter creates its own count and a nested function.
2. nonlocal tells next_count to update count from its enclosing function.
3. Returning next_count preserves access to that state for later calls.

**Common mistake to avoid:** Without nonlocal, assignment to count inside next_count creates a local variable and raises UnboundLocalError.

### Exercise 10 · Recursion

**Task.** Write a recursive function with a base case.

**Why it works.** A base case is essential to terminate recursion.

*Solution status: reviewed/reconstructed reference example.*

````python
# Negative inputs are rejected first using raise.
# The base case n == 0 returns one without making further calls.
def factorial(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n == 0:
        return 1
    return n * factorial(n - 1)

# For positive n, the function calls itself with n - 1 until reaching that base case.
print("Factorial 5:", factorial(5))
````

**Observed output (example run)**

```text
Factorial 5: 120
```

**Step-by-step explanation**

1. Negative inputs are rejected first using raise.
2. The base case n == 0 returns one without making further calls.
3. For positive n, the function calls itself with n - 1 until reaching that base case.

**Common mistake to avoid:** A recursive function without a reachable base case may recurse until RecursionError.

### Exercise 14 · reduce accumulation

**Task.** Combine a sequence into one accumulated result.

**Why it works.** reduce repeatedly combines the accumulator with the next item.

*Solution status: reviewed/reconstructed reference example.*

````python
from functools import reduce

# reduce starts with the initial accumulator value one.
numbers = [2, 3, 4]
# The lambda combines the current total with the next number by multiplication.
product = reduce(lambda x, y: x * y, numbers, 1)
# The final result is a single product, not a list.
print("Product:", product)
````

**Observed output (example run)**

```text
Product: 24
```

**Step-by-step explanation**

1. reduce starts with the initial accumulator value one.
2. The lambda combines the current total with the next number by multiplication.
3. The final result is a single product, not a list.

**Common mistake to avoid:** Using an initial zero with multiplication makes every product zero.

### Exercise 18 · Methods and instance state

**Task.** Update instance attributes through a method.

**Why it works.** self refers to the current instance.

*Solution status: reviewed/reconstructed reference example.*

````python
# __init__ creates account state when an object is instantiated.
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount

    def get_balance(self):
        return self.balance

# deposit checks that the amount is positive before changing self.balance.
account = BankAccount("Sally", 100)
account.deposit(50)
# get_balance returns the updated balance for the caller to print.
print("Owner:", account.owner)
print("Balance:", account.get_balance())
````

**Observed output (example run)**

```text
Owner: Sally
Balance: 150
```

**Step-by-step explanation**

1. __init__ creates account state when an object is instantiated.
2. deposit checks that the amount is positive before changing self.balance.
3. get_balance returns the updated balance for the caller to print.

**Common mistake to avoid:** A method changing instance data uses self to select the correct object.

### Exercise 20 · Method composition

**Task.** Call one instance method from another.

**Why it works.** Instance methods can reuse behavior through self.method().

*Solution status: reviewed/reconstructed reference example.*

````python
# area calculates width times height from the current instance.
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def report(self):
        return f"Area: {self.area()}"

# report calls self.area() instead of duplicating its formula.
rectangle = Rectangle(4, 6)
# The returned formatted string is displayed by print in the caller.
print(rectangle.report())
````

**Observed output (example run)**

```text
Area: 24
```

**Step-by-step explanation**

1. area calculates width times height from the current instance.
2. report calls self.area() instead of duplicating its formula.
3. The returned formatted string is displayed by print in the caller.

**Common mistake to avoid:** Using self.area without parentheses refers to the method object rather than calling it.

### Practical 01 · Class 5 · Task 1: recursive Fibonacci

**Task.** Use three parameters (two with defaults) to calculate the eighth Fibonacci value.

**Source context:** adapted from course practice. This is a reference implementation, not claimed as an original submission.

````python
# The default previous and current values encode consecutive Fibonacci terms.
def fibonacci(position, previous=0, current=1):
    if position < 1:
        raise ValueError("Position must be positive")
    if position == 1:
        return previous
    return fibonacci(position - 1, current, previous + current)

# The base condition stops recursive calls at the requested position.
print("Fibonacci position 8:", fibonacci(8))
# Each recursive call advances the pair, moving toward the requested term.
assert fibonacci(8) == 13
````

**Observed output (example run)**

```text
Fibonacci position 8: 13
```

**Step-by-step explanation**

1. The default previous and current values encode consecutive Fibonacci terms.
2. The base condition stops recursive calls at the requested position.
3. Each recursive call advances the pair, moving toward the requested term.

**Common mistake to avoid:** A recursive call must decrease the remaining position to reach a stopping condition.

### Practical 05 · Class 5 · Task 5: factorial with reduce

**Task.** Compute factorial using reduce, and verify it against the standard library.

**Source context:** adapted from course practice. This is a reference implementation, not claimed as an original submission.

````python
from functools import reduce
from math import factorial
from operator import mul

# range(1,n+1) supplies all factors from one through n.
# reduce with mul and initial one multiplies those factors.
def factorial_reduce(n):
    if n < 0:
        raise ValueError("Factorial is undefined for negative integers")
    return reduce(mul, range(1, n + 1), 1)

# assert compares the result with math.factorial for each test case.
for n in (10, 17, 22, 31):
    computed = factorial_reduce(n)
    assert computed == factorial(n)
    print("Factorial validated for:", n)
````

**Observed output (example run)**

```text
Factorial validated for: 10
Factorial validated for: 17
Factorial validated for: 22
Factorial validated for: 31
```

**Step-by-step explanation**

1. range(1,n+1) supplies all factors from one through n.
2. reduce with mul and initial one multiplies those factors.
3. assert compares the result with math.factorial for each test case.

**Common mistake to avoid:** The identity for an empty multiplication is one, so factorial(0) also works.

### Practical 06 · Class 5 · Task 6: higher-order function

**Task.** Write a helper that accepts an operation and its data, then returns a list.

**Source context:** adapted from course practice. This is a reference implementation, not claimed as an original submission.

````python
# convolve receives operation, a function and an iterable as arguments.
def convolve(operation, function, data):
    return list(operation(function, data))

# It invokes operation(function, data), for example map with a lambda.
groups = [[1, 5, 3], [3, 6], [9, 11, 45, 11], [5, 2, 6, 3, 7]]
result = convolve(map, lambda group: sum(group), groups)
# list materializes the returned iterator before returning it.
print("Convolved sums:", result)
assert result == [9, 9, 76, 23]
````

**Observed output (example run)**

```text
Convolved sums: [9, 9, 76, 23]
```

**Step-by-step explanation**

1. convolve receives operation, a function and an iterable as arguments.
2. It invokes operation(function, data), for example map with a lambda.
3. list materializes the returned iterator before returning it.

**Common mistake to avoid:** Pass the map function object; calling map() without its required arguments would fail.
